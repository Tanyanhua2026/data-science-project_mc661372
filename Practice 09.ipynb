```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv('FlooddataMasterListrev.csv')
print(df.columns)
```

```python
import numpy as np
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.bar(df['Year'], df['Total (M>4)'], label='Large Floods M>4')
plt.bar(df['Year'], df['Total (M>6)'], label='Extreme Floods M>6')
plt.xticks(np.arange(1985,2013,1), rotation=45)
plt.xlabel("Year")
plt.ylabel("Count")
plt.title("Annual Large and Extreme Flood Count")
plt.legend()
plt.grid(axis='y', alpha=0.3)

plt.savefig("flood_count_plot.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
plt.plot(df['Year'], df['Total (M>4)'], marker='o', label='M>4')
plt.plot(df['Year'], df['Total (M>6)'], marker='s', label='M>6')
plt.xticks(np.arange(1985, 2013, 1), rotation=45)

plt.xlabel('Year')
plt.ylabel('Count')
plt.title('Annual Floods (M>4 vs M>6)')
plt.legend()
plt.grid(True)

plt.savefig("annual_flood_line_plot.png", dpi=300, bbox_inches="tight")
plt.show()
```
